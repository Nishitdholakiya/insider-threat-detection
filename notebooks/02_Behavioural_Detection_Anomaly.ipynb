## Is there a shift in AAM0658's pattern?

```python
suspect_logons['hour'] = suspect_logons['date'].dt.hour
print(suspect_logons.groupby(suspect_logons['date'].dt.to_period('M'))['hour'].agg(['mean','min','max']))
```

### Observation: 

January through September (2010-01 to 2010-09): rock-solid consistent

Mean hour: exactly 15.0 every single month
Min: always 9 (9 AM — earliest logon)
Max: always 21 (9 PM — latest logoff)

This user had an extremely stable, predictable routine for 9 straight months. Textbook normal behavior.
October (2010-10): something changes

Mean drops to 13.04 (pulled earlier/more spread out)
Min drops to 0 — meaning at least one logon happened at midnight (hour 0)
Max stays at 21

November (2010-11): continues the shift

Mean at 14.1
Min at 2 — activity as early as 2 AM
Max still 21

What this means
Starting in October, this user began logging on/off at hours (midnight, 2 AM) that never once happened in the previous 9 months. This is precisely the "after-hours activity that didn't exist before" signal from Scenario 1's description — and you found it just by grouping timestamps by month. So behavioral deviation from an established personal baseline is detectable, even with something as simple as monthly hour statistics. This is exactly what autoencoder will later learn to do automatically, at scale, across thousands of users and dozens of features — but signal exists.
